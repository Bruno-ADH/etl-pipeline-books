# Test des connexions

Ce notebook permet de vérifier que tous les services sont accessibles.

```python
# Test connexion MinIO
from minio import Minio

client = Minio(
    endpoint="localhost:9000",
    access_key="minioadmin",
    secret_key="minioadmin123",
    secure=False
)

buckets = [b.name for b in client.list_buckets()]
print(f"✓ MinIO connecté — Buckets : {buckets}")
```

```python
# Test API Open-Meteo
import httpx

response = httpx.get(
    "https://api.open-meteo.com/v1/forecast",
    params={"latitude": 6.3676, "longitude": 2.4252, "current_weather": True}
)
data = response.json()
print(f"✓ Open-Meteo API OK — Météo Cotonou : {data['current_weather']['temperature']}°C")
```

```python
# Test scraping books.toscrape.com
import requests
from bs4 import BeautifulSoup

response = requests.get("https://books.toscrape.com/")
soup = BeautifulSoup(response.text, "lxml")
books = soup.find_all("article", class_="product_pod")
print(f"✓ books.toscrape.com accessible — {len(books)} livres sur la page d'accueil")
```
